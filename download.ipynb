```python
from datasets import load_dataset
import pandas as pd
import os

print("Download e processamento del dataset in corso...")

ds = load_dataset("nezahatkorkmaz/heart-disease-dataset")
df = ds['train'].to_pandas()

# 1. Binarizziamo il target (0 = sano, >0 = malato)
df['num'] = (df['num'] > 0).astype(int)

# 2. Trasformiamo le colonne testuali in numeriche (One-Hot Encoding)
# Le colonne testuali individuate sono: age_group, cholesterol_level, bp_level
df = pd.get_dummies(df, columns=['age_group', 'cholesterol_level', 'bp_level'], drop_first=True, dtype=int)

# 3. Spostiamo il target 'num' per farlo diventare l'ultima colonna
target = df.pop('num') # Rimuove la colonna 'num' e la salva in 'target'
df['num'] = target     # La riaggiunge alla fine del DataFrame

# 4. Salviamo il file pronto per l'uso
os.makedirs("data", exist_ok=True)
df.to_csv("data/heart_huggingface_mod.csv", index=False)

print(f"Fatto! Il nuovo dataset ha {df.shape[1]} colonne (incluso il target).")
```

```python
import pandas as pd

# Caricamento del dataset
file_path = "data/heart_johnsmith88_mod.csv"
df = pd.read_csv(file_path)

print(f"Dimensione originale: {df.shape}")

# Calcolo e rimozione dei duplicati esatti
duplicati = df.duplicated().sum()
print(f"Trovati {duplicati} pazienti duplicati. Rimozione in corso...")

df_clean = df.drop_duplicates()

# Salvataggio del dataset pulito (sovrascriviamo il file per mantenere compatibile il codice)
df_clean.to_csv(file_path, index=False)

print(f"Dimensione finale pulita: {df_clean.shape}")
```
